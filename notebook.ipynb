```python
import pandas as pd
import os
from dotenv import load_dotenv
import requests
from alpaca.trading.client import TradingClient
from alpaca.data.live import StockDataStream
```

CARGA DE API KEY junto a otras variables necesarias

```python
load_dotenv()  # Load environment variables from .env file

API_KEY = os.getenv("API_KEY")
SECRET_KEY = os.getenv("SECRET_KEY")
Mi_URL = os.getenv("ENDPOINT")

#Inicio de sesion de cliente de Alpaca
client = TradingClient(api_key=API_KEY,
                       secret_key=SECRET_KEY, 
                       paper=True
                       )
cuenta = client.get_account()

print(f"Estado de la cuenta {cuenta.status}")
print(f"Efectivo disponible: {cuenta.cash}")
```

```python
wss_client = StockDataStream(api_key=API_KEY, secret_key=SECRET_KEY)

async def handle_trade(trade):
    print(f"Trade update for {trade.symbol}: Price: {trade.price}, Size: {trade.size}, Timestamp: {trade.timestamp}")

wss_client.subscribe_bars(handle_trade, "AAPL")

wss_client = StockDataStream(api_key=API_KEY, secret_key=SECRET_KEY)

async def handle_bars(bar):
    print(f"Bar update for {bar.symbol}: Open: {bar.open}, High: {bar.high}, Low: {bar.low}, Close: {bar.close}, Timestamp: {bar.timestamp}")

wss_client.subscribe_bars(handle_bars, "AAPL")

async def handle_trades(trade):
    print(f"Trade update for {trade.symbol}: Price: {trade.price}, Size: {trade.size}, Timestamp: {trade.timestamp}")

wss_client.subscribe_to_trades(handle_trades, "AAPL")

await wss_client._run()
```
